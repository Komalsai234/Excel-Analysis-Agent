```python
df1 = pd.read_excel("Premium_Metrics_202550.xlsx")
```

```python
df1
```

```python
df1["Function"] = df1["Function"].ffill()
```

```python
df1
```

```python
md_blocks = []

for function, g in df1.groupby("Function"):
    md = f"## {function}\n\n"
    md += g.drop(columns=["Function"]).to_markdown(index=False)
    md_blocks.append(md)

markdown_output = "\n\n".join(md_blocks)
```

```python
print(markdown_output)
```
